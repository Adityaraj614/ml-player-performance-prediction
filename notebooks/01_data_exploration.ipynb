```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
data_path = "../data/synthetic/player_performance.csv"
df = pd.read_csv(data_path)
df.head()
```

```python
df.shape
```

```python

df.dtypes
```

```python

df.describe()
```

```python
# Class distribution
df['performance'].value_counts().sort_index()
```

```python
import matplotlib.pyplot as plt

df['performance'].value_counts().sort_index().plot(kind='bar')
plt.xlabel("Performance Class")
plt.ylabel("Count")
plt.title("Class Distribution")
plt.show()
```

```python
df[['accuracy', 'time_played', 'deaths', 'score']].hist(
    bins=30,
    figsize=(10, 8)
)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
df[['accuracy', 'time_played', 'deaths', 'score']].boxplot()
plt.title("Feature Distributions (Boxplot)")
plt.show()
```

```python
# Correlation matrix
corr = df.corr()

corr
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 6))
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt=".2f")
plt.title("Feature Correlation Heatmap")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(x='performance', y='score', data=df)
plt.title("Score Distribution by Performance Class")
plt.xlabel("Performance Class")
plt.ylabel("Score")
plt.show()
```
